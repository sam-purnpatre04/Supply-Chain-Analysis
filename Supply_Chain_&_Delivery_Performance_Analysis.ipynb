```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
```

```python
df.groupby(df['Order_Date'].dt.to_period('M'))['Total_Revenue'].sum()
```

```python
monthly_revenue = df.groupby(
    df['Order_Date'].dt.to_period('M')
)['Total_Revenue'].sum()

monthly_revenue.plot(kind='line', figsize=(12,5))

plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)
plt.show()
```

```python
df.groupby('Customer_Region')['Total_Revenue'].sum().sort_values().plot(
    kind='barh',
    figsize=(8,5)
)

plt.title('Revenue by Customer Region')
plt.xlabel('Total Revenue')
plt.ylabel('Customer Region')
plt.show()
```

```python
df['Delivery_Status'].value_counts().plot(
    kind='bar',
    figsize=(7,5)
)

plt.title('Order Delivery Status')
plt.xlabel('Delivery Status')
plt.ylabel('Number of Orders')
plt.xticks(rotation=0)
plt.show()
```

```python
df.groupby('Supplier_Name')['Delay_Days'].mean().sort_values().plot(
    kind='barh',
    figsize=(8,5)
)

plt.title('Average Delivery Delay by Supplier')
plt.xlabel('Average Delay (Days)')
plt.ylabel('Supplier')
plt.show()
```

```python
pd.crosstab(
    df['Delivery_Status'],
    df['Return_Status']
).plot(
    kind='bar',
    figsize=(8,5)
)

plt.title('Returns by Delivery Status')
plt.xlabel('Delivery Status')
plt.ylabel('Number of Orders')
plt.xticks(rotation=0)
plt.show()
```

```python
pd.crosstab(
    df['Shipping_Mode'],
    df['Delivery_Status']
).plot(
    kind='bar',
    stacked=True,
    figsize=(9,5)
)

plt.title('Delivery Performance by Shipping Mode')
plt.xlabel('Shipping Mode')
plt.ylabel('Number of Orders')
plt.xticks(rotation=0)
plt.show()
```

```python
df.groupby('Delay_Days')['Customer_Rating'].mean().plot(
    kind='line',
    marker='o',
    figsize=(8,5)
)

plt.title('Customer Rating vs Delivery Delay')
plt.xlabel('Delay Days')
plt.ylabel('Average Customer Rating')
plt.grid(True)
plt.show()
```
